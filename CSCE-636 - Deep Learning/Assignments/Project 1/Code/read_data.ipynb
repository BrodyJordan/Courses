```python
import pickle

training_data = pickle.load(open("DS-15-samples_n_k_m_P", "rb"))
training_targets = pickle.load(open("DS-15-samples_mHeights", "rb"))

print("Training data shape:", len(training_data), "samples with", len(training_data[0]), "features each")
print("Training targets shape:", len(training_targets), "samples with 1 features each")

# training data in shape [n,k,m,P], where n = 9, k in {4, 5, 6}, m in {2, 3, ..., n - k}, P 2 dimensional numpy array of size k by n-k
# count how many of each of the 9 combinations of n, k, m are present in the training data

from collections import defaultdict
combination_counts = defaultdict(int)
for sample in training_data:
    n = sample[0]
    k = sample[1]
    m = sample[2]
    combination_counts[(n, k, m)] += 1
for combo, count in combination_counts.items():
    print(f"Combination n={combo[0]}, k={combo[1]}, m={combo[2]}: {count} samples")

# sum them

```
